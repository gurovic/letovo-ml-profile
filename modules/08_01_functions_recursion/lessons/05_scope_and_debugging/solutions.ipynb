# Решения: scope и отладка

**Для преподавателя.** Все задачи из `lesson.ipynb` и `homework.ipynb`.

```python
PREDICTIONS = [1, 0, 1, 1, 0, 1, 1, 0, 1, 0]
LABELS =      [1, 0, 1, 0, 0, 1, 1, 0, 1, 1]
SCORES = [72, 55, 88, 44, 61, 90, 77, 48, 83, 58]
```

## Урок. 1–2. accuracy / my_accuracy

```python
def accuracy(preds, labels):
    if len(preds) != len(labels):
        return None
    correct = sum(p == y for p, y in zip(preds, labels))
    return correct / len(labels)


REF_ACCURACY = accuracy(PREDICTIONS, LABELS)


def my_accuracy(preds, labels):
    if len(preds) != len(labels):
        return None
    if not preds:
        return 0.0
    correct = 0
    for p, y in zip(preds, labels):
        if p == y:
            correct += 1
    return correct / len(labels)


assert abs(my_accuracy(PREDICTIONS, LABELS) - REF_ACCURACY) < 1e-9
```

## Урок. 3. accuracy_fixed

```python
def accuracy_fixed(preds, labels):
    correct = 0
    for i in range(len(preds)):
        if preds[i] == labels[i]:
            correct += 1
    return correct / len(preds) if preds else 0.0
```

## Урок. 4. count_correct

```python
def count_correct(preds, labels):
    total = 0
    for p, y in zip(preds, labels):
        if p == y:
            total += 1
    return total


assert count_correct(PREDICTIONS, LABELS) == 8
```

## Урок. 5. predict_pass

```python
def predict_pass(score, threshold=60):
    return 1 if score >= threshold else 0


assert predict_pass(72) == 1
```

## Урок. 6–7. confusion_counts

На всех 10 объектах: `(5, 1, 1, 3)`; на первых 4: `(2, 1, 0, 1)`. Проверка: `tp + tn == count_correct == 8`.

```python
def confusion_counts(preds, labels):
    tp = 0
    fp = 0
    fn = 0
    tn = 0
    for i in range(len(preds)):
        p = preds[i]
        y = labels[i]
        if p == 1 and y == 1:
            tp += 1
        elif p == 1 and y == 0:
            fp += 1
        elif p == 0 and y == 1:
            fn += 1
        else:
            tn += 1
    return tp, fp, fn, tn


assert confusion_counts(PREDICTIONS, LABELS) == (5, 1, 1, 3)
assert confusion_counts(PREDICTIONS[:4], LABELS[:4]) == (2, 1, 0, 1)
```

## ДЗ. 2. batch_predict_pass

```python
def batch_predict_pass(scores, threshold=60):
    return [predict_pass(s, threshold) for s in scores]


preds = batch_predict_pass(SCORES, 60)
assert len(preds) == len(SCORES)
```

## ДЗ. 4. shadowing

```python
def accuracy_shadow_bug(preds, labels):
    labels = 0  # затёрли аргумент!
    correct = sum(p == labels for p in preds)
    return correct / len(preds)


def accuracy_shadow_ok(preds, labels):
    correct = sum(p == y for p, y in zip(preds, labels))
    return correct / len(preds)


assert abs(accuracy_shadow_ok(PREDICTIONS, LABELS) - REF_ACCURACY) < 1e-9
```

## ДЗ. 5. precision и recall (по желанию)

precision = 5 / (5 + 1) ≈ 0.833; recall = 5 / (5 + 1) ≈ 0.833. При нулевом знаменателе договариваемся возвращать `None`.

```python
def precision(tp, fp):
    if tp + fp == 0:
        return None
    return tp / (tp + fp)


def recall(tp, fn):
    if tp + fn == 0:
        return None
    return tp / (tp + fn)


tp, fp, fn, tn = confusion_counts(PREDICTIONS, LABELS)
assert abs(precision(tp, fp) - 5 / 6) < 1e-9
assert abs(recall(tp, fn) - 5 / 6) < 1e-9
```
