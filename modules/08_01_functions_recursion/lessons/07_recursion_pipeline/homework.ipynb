## 5. Мини-pipeline с lambda

Используя строку `TEXT`, соберите pipeline из шагов:
`strip -> lower -> split -> оставить только токены длиной >= 6`.

Нужно описать список функций `text_pipeline`, применить его через `apply_pipeline` и получить список длинных слов.
После решения проверьте результат по `assert`.


```python
def apply_pipeline(data, steps):
    result = data
    for step in steps:
        result = step(result)
    return result


text_pipeline = [
    lambda s: s.strip().lower(),
    lambda s: s.split(),
    lambda tokens: [t for t in tokens if len(t) >= 6],
]

result = apply_pipeline(TEXT, text_pipeline)
# assert result == ["recursion", "nested", "structures", "python"]
print(result)
```
