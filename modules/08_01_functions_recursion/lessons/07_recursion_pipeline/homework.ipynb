## 5. Мини-pipeline с lambda

Соберите pipeline: `strip -> lower -> split -> оставить токены длиной >= 5`.


```python
def apply_pipeline(data, steps):
    result = data
    for step in steps:
        result = step(result)
    return result


text_pipeline = [
    lambda s: s.strip().lower(),
    lambda s: s.split(),
    lambda tokens: [t for t in tokens if len(t) >= 5],
]

result = apply_pipeline(TEXT, text_pipeline)
# assert result == ["recursion", "helps", "nested", "structures", "python"]
print(result)
```
