# ДЗ: площадь и история ошибки

```python
import numpy as np

def trapezoid_from_values(values, step=1.0):
    values = np.asarray(values, dtype=float)
    return float(step * (0.5 * values[0] + np.sum(values[1:-1]) + 0.5 * values[-1]))
```
