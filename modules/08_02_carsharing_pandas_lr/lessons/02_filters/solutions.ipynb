# Решения. Фильтры — урок и ДЗ

Эталон к ноутбуку урока и к домашнему заданию. Ученикам до сдачи не показывать. Каждая задача стоит в своей ячейке, в том же порядке, что у ученика.


```python
from pathlib import Path
import pandas as pd


LISTINGS_URL = (
    "https://raw.githubusercontent.com/gurovic/letovo-ml-profile/main/"
    "modules/08_02_carsharing_pandas_lr/data/listings.csv"
)


def find_listings_csv():
    for p in (
        Path("listings.csv"),
        Path("../listings.csv"),
        Path("../../data/listings.csv"),
        Path("../../../data/listings.csv"),
        Path("../data/listings.csv"),
    ):
        if p.exists():
            return p.resolve()
    return LISTINGS_URL


LISTINGS_PATH = find_listings_csv()
df = pd.read_csv(LISTINGS_PATH)
```

## Урок


## 1. Фильтр по району


```python
center = df[df['neighbourhood'] == 'center']
print('center:', len(center), 'из', len(df))
```

## 2. Вместительные объявления


```python
large_listings = df[df['accommodates'] >= 6]
print(large_listings[['listing_id', 'accommodates', 'price']].head())
```

## 3. Составная маска


```python
center_long = df[(df['neighbourhood'] == 'center') & (df['accommodates'] >= 4)]
print(len(center_long))
```

## 4. Отрицание


```python
not_center = df[~(df['neighbourhood'] == 'center')]
assert len(not_center) + len(center) == len(df)
print('not center:', len(not_center))
```

## 5. Сортировка и топ-N


```python
top3 = df.sort_values('price', ascending=False).head(3)
print(top3[['listing_id', 'price']])
```

## 6. Доля фильтра


```python
share_center = round(len(center) / len(df), 3)
print('доля center:', share_center)
```

## 7. Первая строка после фильтра


```python
first_center_id = str(center.iloc[0]['listing_id'])
print(first_center_id)
```

## 8. Тип number_of_reviews


```python
reviews = df['number_of_reviews'].astype(int)
print(int(reviews.min()), int(reviews.max()))
```

## ДЗ


## 1. Entire


```python
n_entire = int((df['room_type'] == 'entire').sum())
print(n_entire)
```

## 2. Отзывы от 10 до 50


```python
n_mid_reviews = int(((df['number_of_reviews'] >= 10) & (df['number_of_reviews'] <= 50)).sum())
print(n_mid_reviews)
```

## 3. Топ-5 по цене


```python
top5 = df.sort_values('price', ascending=False).head(5)
print(top5[['listing_id', 'price']])
```

## 4. Доля центра и короткая оценка


```python
share_center = round((df['neighbourhood'] == 'center').mean(), 3)
SHARE_NOTE = f'center ≈ {share_center:.0%} объявлений — ориентир доли центра'
print(share_center, SHARE_NOTE)
```

## 5. Не entire


```python
n_not_entire = int((df['room_type'] != 'entire').sum())
print(n_not_entire)
```

## Самостоятельно


## 9. Пляж и первые строки


```python
beach_head = (
    df[(df['neighbourhood'] == 'beach') & (df['accommodates'] >= 4)]
    .head(3)
)
assert len(beach_head) == 3
assert (beach_head['neighbourhood'] == 'beach').all()
assert (beach_head['accommodates'] >= 4).all()
print(beach_head[['listing_id', 'accommodates', 'price']])
```

## 10. Сводка по дорогим объявлениям


```python
expensive = df[df['price'] >= 120]
summary = expensive.describe()
expensive_min = summary.loc['min', 'price']
expensive_mean = summary.loc['mean', 'price']
assert float(expensive_min) >= 120
assert float(expensive_mean) >= float(expensive_min)
print(expensive_min, expensive_mean)
```

## 11. Первая строка вне beach


```python
not_beach = df[~(df['neighbourhood'] == 'beach')]
first_not_beach_id = str(not_beach.iloc[0]['listing_id'])
assert isinstance(first_not_beach_id, str) and first_not_beach_id.startswith('L')
print(first_not_beach_id)
```

## 12. Два самых дорогих в old_town


```python
old_town = (
    df[df['neighbourhood'] == 'old_town']
    .sort_values('price', ascending=False)
    .head(2)
)
old_town_top = old_town.loc[:, ['listing_id', 'price']]
assert len(old_town_top) == 2
assert list(old_town_top.columns) == ['listing_id', 'price']
assert old_town_top['price'].is_monotonic_decreasing
print(old_town_top)
```

## 13. info по срезу вне центра


```python
busy_outside = df[
    (df['number_of_reviews'] > 200) & ~(df['neighbourhood'] == 'center')
]
busy_outside.info()
n_busy_outside = len(busy_outside)
assert isinstance(n_busy_outside, int)
assert 0 < n_busy_outside < len(df)
print(n_busy_outside)
```
