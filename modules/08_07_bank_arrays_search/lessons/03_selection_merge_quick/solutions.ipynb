```python
from pathlib import Path
import time
import pandas as pd


def _find(name: str) -> Path:
    for p in (Path(name), Path(f'../../data/{name}'), Path(f'../data/{name}')):
        if p.exists():
            return p.resolve()
    raise FileNotFoundError(f'{name} не найден рядом с ноутбуком')


unsorted_df = pd.read_csv(_find('bank_transactions_unsorted.csv'))
by_id_df = pd.read_csv(_find('bank_transactions_sorted_by_txn_id.csv'))
by_amount_df = pd.read_csv(_find('bank_transactions_sorted_by_amount.csv'))
tiny_df = pd.read_csv(_find('bank_transactions_tiny.csv'))

unsorted_txns = list(unsorted_df[['txn_id', 'amount', 'day', 'risk_score']].itertuples(index=False, name=None))
id_txns = list(by_id_df[['txn_id', 'amount', 'day', 'risk_score']].itertuples(index=False, name=None))
amount_txns = list(by_amount_df[['txn_id', 'amount', 'day', 'risk_score']].itertuples(index=False, name=None))
id_list = [t[0] for t in id_txns]
amount_list = [t[1] for t in amount_txns]
```

```python
def selection_sort(nums):
    arr = nums[:]
    n = len(arr)
    for i in range(n):
        m = i
        for j in range(i + 1, n):
            if arr[j] < arr[m]:
                m = j
        arr[i], arr[m] = arr[m], arr[i]
    return arr


def merge_sorted(a, b):
    i = 0
    j = 0
    out = []
    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            out.append(a[i])
            i += 1
        else:
            out.append(b[j])
            j += 1
    out.extend(a[i:])
    out.extend(b[j:])
    return out


def merge_sort(nums):
    if len(nums) <= 1:
        return nums[:]
    mid = len(nums) // 2
    left = merge_sort(nums[:mid])
    right = merge_sort(nums[mid:])
    return merge_sorted(left, right)


def partition_once(nums):
    pivot = nums[len(nums) // 2]
    left = [x for x in nums if x < pivot]
    same = [x for x in nums if x == pivot]
    right = [x for x in nums if x > pivot]
    return left + same[:-1], pivot, right


arr = [9, 1, 5, 3, 7]
sorted_arr = selection_sort(arr)
merged = merge_sorted([1, 4, 9], [2, 3, 7])
test = [12, 5, 1, 8, 3, 7]
left, pivot, right = partition_once([8, 2, 9, 4, 6])
SORT_NOTE = (
    'Selection sort делает много сравнений и перестановок, поэтому растёт как O(n^2). '
    'Mergesort делит массив и сливает отсортированные части, что обычно даёт O(n log n).'
)
part = [x[1] for x in unsorted_txns[:30]]
sorted_part = selection_sort(part)
vals = [x[1] for x in unsorted_txns[:120]]
res = merge_sort(vals)
QUICK_NOTE = (
    'Quicksort может замедляться на почти отсортированных данных при неудачном pivot: '
    'разбиения становятся неравномерными, и глубина рекурсии растёт.'
)
print(sorted_arr)
print(merged)
print(merge_sort(test))
print(left, pivot, right)
```
