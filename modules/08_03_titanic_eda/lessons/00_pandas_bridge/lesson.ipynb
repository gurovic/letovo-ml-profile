```python
from pathlib import Path
import pandas as pd

TITANIC_URL = (
    "https://raw.githubusercontent.com/gurovic/letovo-ml-profile/main/"
    "modules/08_03_titanic_eda/data/titanic.csv"
)

def find_titanic_csv():
    for p in (
        Path("titanic.csv"),
        Path("../data/titanic.csv"),
        Path("../../data/titanic.csv"),
        Path("../../../data/titanic.csv"),
    ):
        if p.exists():
            return p.resolve()
    return TITANIC_URL

TITANIC_PATH = find_titanic_csv()
df = pd.read_csv(TITANIC_PATH)
```
