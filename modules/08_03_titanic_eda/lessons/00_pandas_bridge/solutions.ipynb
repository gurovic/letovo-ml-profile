# Решения. Та же pandas на новой таблице

Эталон перенесённой пары. Ученикам до сдачи не показывать.


```python
from pathlib import Path
import pandas as pd

TITANIC_URL = (
    "https://raw.githubusercontent.com/gurovic/letovo-ml-profile/main/"
    "modules/08_03_titanic_eda/data/titanic.csv"
)

def find_titanic_csv():
    for p in (
        Path("titanic.csv"),
        Path("../data/titanic.csv"),
        Path("../../data/titanic.csv"),
        Path("../../../data/titanic.csv"),
    ):
        if p.exists():
            return p.resolve()
    return TITANIC_URL

TITANIC_PATH = find_titanic_csv()
df = pd.read_csv(TITANIC_PATH)
```

## 1. Размер новой таблицы


```python
n_rows = int(len(df))
n_cols = int(df.shape[1])
TABLE_NOTE = f'В таблице {n_rows} строк и {n_cols} столбцов, каждая строка — пассажир.'
assert str(n_rows) in TABLE_NOTE
print(TABLE_NOTE)
```

## 2. В каком столбце больше всего пустых ячеек


```python
missing = df.isna().sum()
gap_col = str(missing.idxmax())
n_gap = int(missing.max())
GAP_NOTE = f'Больше всего пустых ячеек в столбце {gap_col}: их {n_gap}.'
assert gap_col in GAP_NOTE and str(n_gap) in GAP_NOTE
print(GAP_NOTE)
```

## 3. Пассажиры первого класса


```python
first = df.loc[df['pclass'] == 1]
n_first = int(len(first))
median_fare_first = float(first['fare'].median())
FIRST_NOTE = (
    f'Пассажиров первого класса {n_first}, медиана стоимости билета {median_fare_first:.1f}.'
)
assert str(n_first) in FIRST_NOTE
print(FIRST_NOTE)
```

## 4. Кого в таблице больше


```python
sex_counts = df['sex'].value_counts()
top_sex = str(sex_counts.idxmax())
n_top_sex = int(sex_counts.max())
SEX_NOTE = f'Чаще в таблице встречается {top_sex}: таких строк {n_top_sex}.'
assert top_sex in SEX_NOTE and str(n_top_sex) in SEX_NOTE
print(SEX_NOTE)
```

## 5. Средняя стоимость билета по классу


```python
mean_fare = df.groupby('pclass')['fare'].mean()
best = int(mean_fare.idxmax())
FARE_NOTE = f'Выше всех средняя стоимость билета у класса {best}: {float(mean_fare.max()):.1f}.'
assert len(FARE_NOTE) > 20
print(mean_fare)
print(FARE_NOTE)
```

## 6. Женщины третьего класса


```python
women = df.loc[(df['sex'] == 'female') & (df['pclass'] == 3)]
n_women_third = int(len(women))
n_women_third_age_na = int(women['age'].isna().sum())
median_age_women_third = float(women['age'].median())
WOMEN_NOTE = (
    f'Женщин третьего класса {n_women_third}. '
    f'Возраст не указан у {n_women_third_age_na} из них, '
    f'медиана указанного возраста {median_age_women_third:.1f}.'
)
assert str(n_women_third) in WOMEN_NOTE and str(n_women_third_age_na) in WOMEN_NOTE
print(WOMEN_NOTE)
```

## ДЗ. Дети в таблице пассажиров


```python
from pathlib import Path
import pandas as pd

TITANIC_URL = (
    "https://raw.githubusercontent.com/gurovic/letovo-ml-profile/main/"
    "modules/08_03_titanic_eda/data/titanic.csv"
)

def find_titanic_csv():
    for p in (
        Path("titanic.csv"),
        Path("../data/titanic.csv"),
        Path("../../data/titanic.csv"),
        Path("../../../data/titanic.csv"),
    ):
        if p.exists():
            return p.resolve()
    return TITANIC_URL

TITANIC_PATH = find_titanic_csv()
df = pd.read_csv(TITANIC_PATH)
```

## 1. Сколько детей и в каких классах


```python
children = df.loc[df['who'] == 'child']
n_child = int(len(children))
child_by_class = children['pclass'].value_counts()
top_class = int(child_by_class.idxmax())
CHILD_NOTE = f'Детей в таблице {n_child}, больше всего в классе {top_class}.'
assert str(n_child) in CHILD_NOTE
print(CHILD_NOTE)
```

## 2. Возраст детей и пропуски


```python
children = df.loc[df['who'] == 'child']
n_child_age_na = int(children['age'].isna().sum())
median_child_age = float(children['age'].median())
AGE_NOTE = (
    f'Возраст не указан у {n_child_age_na} детей, '
    f'медиана указанного возраста {median_child_age:.1f}.'
)
assert str(n_child_age_na) in AGE_NOTE
print(AGE_NOTE)
```
